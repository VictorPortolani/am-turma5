##Análise de Anomalias de Dados [Census]
**Enunciado: Auditoria e Higienização de Dados Censitários**

Vamos analizar e estruturar a ingestão de um novo pipeline de dados sociodemográficos focado no desenvolvimento de modelos de inteligência artificial. O arquivo **census.csv** foi extraído de um sistema legado, mas o processo automatizado falhou antes mesmo da etapa de modelagem estatística. Como futuros cientistas da computação e engenheiros de dados, a missão de vocês é realizar uma auditoria rigorosa neste arquivo para garantir a integridade da informação.

Dicas de Investigação e Solução

* **Identificação de Linhas Infiltradas**: Ao forçarem o carregamento inicial (pulando as linhas defeituosas do topo), observem o comportamento dos tipos de dados informados pelo `.dtypes`. Se colunas que deveriam ser numéricas (`age` ou `hour-per-week`) foram lidas como `object`, existem *strings* perdidas no meio da massa de dados. O cabeçalho e fragmentos de texto corrompido estão repetidos ao longo do arquivo.
* **Inconsistência Categórica e Caracteres de Escape:** Analisem a distribuição de classes na coluna `income` através do `.value_counts()`. A mesma classe de renda foi acidentalmente dividida em múltiplas categorias distintas devido a falhas de exportação com caracteres de escape textuais (como `\<=50K` versus `<=50K`).
* **Dados Faltantes Não Convencionais:** Avaliem como a ausência de informação foi mascarada pelo sistema de origem. Em bases censitárias clássicas, campos em branco (como `workclass` e `native-country`) frequentemente não aparecem como `NaN` estruturados, mas são substituídos por marcadores categóricos arbitrários, como o caractere `?`.
* **Tratamento de Espaçamento:** A separação por vírgulas pode conter espaços em branco residuais no início das palavras, prejudicando filtros e agrupamentos futuros. O parâmetro `skipinitialspace=True` na leitura do Pandas é um atalho eficiente para padronizar as bordas do texto.

Normalização de Estudantes para `workclass` e `ocupation` assumindo os valores que estavam com `?` para valores `Dependent` e `Student`.

```python
def  normalizarEstudantes(df):
  condition_age_24 = df['age'] < 24
  df.loc[condition_age_24 & (df['workclass'] == ' ?'), 'workclass'] = 'Dependent'
  df.loc[condition_age_24 & (df['occupation'] == ' ?'), 'occupation'] = 'Student'
  return df
base_census = normalizarEstudantes(base_census)
```

```python
base_census.loc[base_census['age'] < 18]
```

```python
base_census.loc[((base_census == ' ?').any(axis=1)) & (base_census['age'] >= 62)]
```

- Normalização de Idosos para `workclass` e `ocupation` assumindo os valores que estavam com `?` para valores `Retired` e `None`.
- Normalização de Desempregados para `workclass` e `ocupation` assumindo os valores que estavam com `?` para valores `Unemployed` e `None`.

```python
def normalizarCenso(df):
    # 1. Tratar Aposentados (62 anos ou mais)
    condicao_aposentado = df['age'] >= 62
    df.loc[condicao_aposentado & (df['workclass'] == ' ?'), 'workclass'] = ' Retired'
    df.loc[condicao_aposentado & (df['occupation'] == ' ?'), 'occupation'] = ' None'

    # 2. Tratar Desempregados (Entre 24 e 61 anos)
    condicao_desempregado = (df['age'] >= 24) & (df['age'] < 62)
    df.loc[condicao_desempregado & (df['workclass'] == ' ?'), 'workclass'] = ' Unemployed'
    df.loc[condicao_desempregado & (df['occupation'] == ' ?'), 'occupation'] = ' None'

    return df

base_census = normalizarCenso(base_census)
```

### Tratamento de `native-country`

```python
print(base_census['native-country'].value_counts())
```

```python
paises_por_etnia = base_census[base_census['native-country'] != ' ?'].groupby('race')['native-country'].agg(lambda x: x.mode()[0])

print(paises_por_etnia)
```

```python
# Verifica a distribuição de renda das pessoas SEM país declarado (os 583 casos)
# O normalize=True transforma os valores em porcentagem
renda_sem_pais = base_census[base_census['native-country'] == ' ?']['income'].value_counts(normalize=True) * 100

# Verifica a distribuição de renda das pessoas COM país declarado (o resto do dataset)
renda_com_pais = base_census[base_census['native-country'] != ' ?']['income'].value_counts(normalize=True) * 100

print("Distribuição de renda - SEM país declarado:")
print(renda_sem_pais.round(2))
print("\nDistribuição de renda - COM país declarado:")
print(renda_com_pais.round(2))
```

```python
# Verifica a distribuição de renda no dataset inteiro
renda_total = base_census['income'].value_counts(normalize=True) * 100

print("Distribuição de renda - Dataset Inteiro:")
print(renda_total.round(2))
```

```python
print(base_census['income'].value_counts())
#Verificação se a renda não tem anomalia (Status: OK)
```

```python
# Filtra apenas o grupo com país ' ?'
sem_pais = base_census[base_census['native-country'] == ' ?']

print("--- Raça dos não declarados ---")
print(sem_pais['race'].value_counts())

print("\n--- Classe de trabalho dos não declarados ---")
print(sem_pais['workclass'].value_counts())
```

As anomalias de países não podem ser excluidas pois os dados são os seguintes, asiátios no total era um número baixo, mas nesses em que o `?` aparece, asiáticos sobem para `14.24%` e negros para `16.47%`, ou seja esses números representam minorias altas, ocasionalmente interferindo no aprendizado da máquina, solução mais propripcía substituir os países em que aparece `?` para `Unkdowm`.

```python
base_census['native-country'] = base_census['native-country'].replace(' ?', ' Unknown')
```
